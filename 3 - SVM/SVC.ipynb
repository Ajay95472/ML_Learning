```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
## Lets create synthetic data points
from sklearn.datasets import make_classification
```

```python
X,y=make_classification(n_samples=1000,n_features=2,n_classes=2,
                        n_clusters_per_class=2,n_redundant=0)
```

```python
pd.DataFrame(X)
```

```python
y
```

```python
X[0]
```

```python
sns.scatterplot(x = pd.DataFrame(X)[0],y = pd.DataFrame(X)[1],hue=y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=10)
```

```python
from sklearn.svm import SVC

svc=SVC(kernel='linear')
```

```python
svc.fit(X_train,y_train)
```

```python
y_pred = svc.predict(X_test)
```

```python
svc.coef0
```

```python
from sklearn.metrics import classification_report,confusion_matrix

print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

## Kernals

```python
rbf=SVC(kernel='rbf')

rbf.fit(X_train,y_train)

## Prediction
y_pred1=rbf.predict(X_test)

print(classification_report(y_test,y_pred1))
print(confusion_matrix(y_test,y_pred1))
```

```python
polynomial=SVC(kernel='poly')

polynomial.fit(X_train,y_train)

## Prediction
y_pred2=polynomial.predict(X_test)

print(classification_report(y_test,y_pred2))
print(confusion_matrix(y_test,y_pred2))
```

```python
sigmoid=SVC(kernel='sigmoid')

sigmoid.fit(X_train,y_train)

## Prediction
y_pred3=sigmoid.predict(X_test)

print(classification_report(y_test,y_pred3))
print(confusion_matrix(y_test,y_pred3))
```

## Hyperparametere Tuning With SVC

```python
from sklearn.model_selection import GridSearchCV
 
# defining parameter range
param_grid = {'C': [0.1, 1, 10, 100, 1000],
              'gamma': [1, 0.1, 0.01, 0.001, 0.0001],
              'kernel': ['rbf']}
```

## C parameter
In SVM, C is one of the most important hyperparameters, especially when you use GridSearchCV.

##### Small C → "I'm okay with some mistakes if I can keep a wider, simpler margin." 
##### Large C → "I really don't want training mistakes, even if I need a more complex boundary."

##### C = 0.01 then SVM says Don't worry too much about individual mistakes. Give me a wider margin.
##### C = 100 then SVM says I strongly want to avoid training mistakes.


## Gamma

#### Gamma is mainly important when you use a non-linear kernel such as RBF
#### High Gamma value = only near by data points are consider
#### Low Gamma value = Far away data points also consider from the desision boundry

```python
grid=GridSearchCV(SVC(),param_grid=param_grid,refit=True,cv=5,verbose=3)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
## Prediction
y_pred4=grid.predict(X_test)
print(classification_report(y_test,y_pred4))
print(confusion_matrix(y_test,y_pred4))
```
